```python
import numpy as np
import os

file_path = "/DATA2/Vivek/Vivekk/Data/NuScenes/nuscenes/samples/LIDAR_TOP/n008-2018-08-01-15-16-36-0400__LIDAR_TOP__1533151061547455.pcd.bin"

# 1. Basic file info
file_size = os.path.getsize(file_path)
print(f"File size (bytes): {file_size}")

# 2. Try reading as uint8 (typical for labels)
data_uint8 = np.fromfile(file_path, dtype=np.uint8)
print("\n[uint8 interpretation]")
print(f"Shape: {data_uint8.shape}")
print(f"Unique values (first 50): {np.unique(data_uint8)[:50]}")
print(f"Max value: {data_uint8.max()}, Min value: {data_uint8.min()}")

# 3. Try reading as float32 (typical for LiDAR points)
data_float32 = np.fromfile(file_path, dtype=np.float32)
print("\n[float32 interpretation]")
print(f"Shape: {data_float32.shape}")

# Check if divisible by 4 or 5 (x,y,z,intensity,(ring))
for dim in [3, 4, 5]:
    if data_float32.shape[0] % dim == 0:
        print(f"Possible point cloud with {dim} features per point: {data_float32.shape[0] // dim} points")

# 4. Print first few entries
print("\nFirst 20 uint8 values:", data_uint8[:20])
print("First 20 float32 values:", data_float32[:20])
```

```python
import numpy as np

file_path = "/DATA2/Vivek/Vivekk/Data/NuScenes/nuscenes/samples/LIDAR_TOP/n008-2018-08-01-15-16-36-0400__LIDAR_TOP__1533151061547455.pcd.bin"

points = np.fromfile(file_path, dtype=np.float32).reshape(-1, 5)

print("Shape:", points.shape)

# Split components
x = points[:, 0]
y = points[:, 1]
z = points[:, 2]
intensity = points[:, 3]
ring = points[:, 4]

print("\nStats:")
print("X range:", x.min(), x.max())
print("Y range:", y.min(), y.max())
print("Z range:", z.min(), z.max())
print("Intensity range:", intensity.min(), intensity.max())
print("Ring unique values:", np.unique(ring)[:10])
```

```python
import numpy as np
import open3d as o3d

# Load point cloud
file_path = "/DATA2/Vivek/Vivekk/Data/NuScenes/nuscenes/samples/LIDAR_TOP/n008-2018-08-01-15-16-36-0400__LIDAR_TOP__1533151061547455.pcd.bin"

points = np.fromfile(file_path, dtype=np.float32).reshape(-1, 5)

xyz = points[:, :3]
intensity = points[:, 3]

# Normalize intensity to [0,1]
intensity_norm = (intensity - intensity.min()) / (intensity.max() - intensity.min() + 1e-8)

# Convert intensity to grayscale colors
colors = np.stack([intensity_norm, intensity_norm, intensity_norm], axis=1)

# Create Open3D point cloud
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(xyz)
pcd.colors = o3d.utility.Vector3dVector(colors)

# Visualize
o3d.visualization.draw_geometries([pcd])
```
